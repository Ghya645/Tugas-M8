```python
#soal 5
departemen = ["Statbis", "Statistika", "Matematika"]
angkatan = [2025, 2026]
for d in departemen:
  for a in angkatan:
    print(f"{d} angkatan {a}")
```

```python
#soal 6
pelamar = [("Budi", 3.2, 5), ("Siti", 3.6, 3), ("Dodi", 3.8, 1)]
for nama, ipk, semester in pelamar:
  if ipk >= 3.5 and semester > 2:
    status = "Diterima"
  else:
    status = "Ditolak"
  print(f"Pelamar: {nama}, Status: {status}")
```

```python
#soal 7
karyawan = [("Joko", "IT", 7000), ("Rina", "HR", 5000), ("Toni", "Data", 8000), ("Lia", "IT", 6500)]
total_gaji = 0
for nama, divisi, gaji in karyawan:
  if divisi == "IT" or divisi == "Data":
    total_gaji += gaji
print(f"Total gaji karyawan: Rp{total_gaji}")
```

```python
#soal 8
penjualan = [("Kopi", 60, False), ("Teh", 30, False), ("Boba", 80, True)]
for jenis, penjualan, promo in penjualan:
  if penjualan > 50 and not promo:
    print(jenis)
```
